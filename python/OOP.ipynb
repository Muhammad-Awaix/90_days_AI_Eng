## Object Oriented Programming

```python
class Car:
    def __init__(self, model:str, year:int, color:str, for_sale:bool):
        self.model=model
        self.year=year
        self.color=color
        self.for_sale=for_sale

    def description(self):
        sale_status = 'available for sale' if self.for_sale else 'not for sale'
        return f"The {self.model} with {self.color} color of {self.year} is {sale_status}"

    def drive(self):
        print(f"You are driving {self.model} Car.")

    def stop(self):
        print(f"You stop the car.")

#Creating Objects
car1 = Car('Mestange', 2021, 'Red', False)
car2 = Car('Toyota', 2019, 'White', True)
car3 = Car('Audi', 2020, 'Gray', True)
#Calling functions and attributes
car1.model
print(car1.description())
car1.drive()
print(car3.description())
```

```python
class Student:
    year = 2021
    no_of_students = 0

    def __init__(self, name:str, age:int, department:str):
        self.name=name
        self.age=age
        self.department=department
        Student.no_of_students+=1

    def details(self):
        print(f"A {self.department} Student name {self.name} {self.age} years old.")
        
s1=Student('Awais', 26, 'CS')
s2=Student('Hamza', 24, 'IT')
s3=Student('Qadeer', 26, 'SE')
print(s1.year)
s1.details()
print(Student.no_of_students)
```

```python
# Calling within the class
class Math:
    def subtract(self, i, j):
        print(i-j)
    def test(self):
        return self.subtract(10,20)
m1 = Math()
m1.test()
```

```python
class Person:
    name='Awais' #Class variable
    def __init__(self,age:int): # This is a constructor, it is called when an instance of the class is created.
        self.age = age #Instance variable
    def greeting(self):
        print("Hello", self.name, "You are", self.age, "years old")
p1 = Person(26)
p1.greeting()
p2 = Person.name="Ali"
p2 = Person(30)
p2.greeting()
p1.greeting() #name is shared between all instances of the class, so changing it for one instance will change it for all instances.
```

## Inheritance

```python
#Relationship b/w them e.g Parent child  shape----> Rectangle, Circle, Triangle
class Shape: #Super Class
    def __init__(self):
        self.color = (0,0,0)
class Rectangle(Shape): #Sub Class
    def __init__(self, w, h):
        Shape.__init__(self) #This is for adding the color attribute in rectangle class
        self.width = w
        self.height = h
    def area(self):
        return self.width*self.height

r1 = Rectangle(10,5)
print(r1.height)
print("The Total Area of a Rectangle is", r1.area())
print(r1.color)
```
