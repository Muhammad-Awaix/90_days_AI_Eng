## Object Oriented Programming

```python
class Math:
    def subtract(self, i, j):
        return i-j
    def test(self):
        print(self.subtract(10,20))
```

```python
class Person:
    name='Awais'
    def __init__(self,s):
        self.name = s
    def greeting(self):
        print("Hello", self.name)
p1 = Person('Ali')
p1.greeting()
```
